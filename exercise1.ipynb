Q1. Reverse String & Every Second Character

```python

# Q1: String Operations

def string_operations(text):
    reversed_text = text[::-1]
    second_chars = text[1::2]
    return reversed_text, second_chars

result = string_operations("abcdef")

print("Reversed String:", result[0])
print("Every Second Character:", result[1])
```

Q2. Squares of Even Numbers

```python

# Q2: Squares of Even Numbers

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_squares = [x**2 for x in numbers if x % 2 == 0]

print("Squares of Even Numbers:", even_squares)
```

Q3. Character Frequency Using Dictionary

```python

# Q3: Character Frequency

def char_frequency(text):
    frequency = {}

    for char in text:
        frequency[char] = frequency.get(char, 0) + 1

    return frequency

print(char_frequency("hello"))
```

Q4. Combine Two Lists Using zip()

```python

# Q4: Combine Lists Using zip()

def combine_lists(keys, values):
    return dict(zip(keys, values))

names = ["Ali", "Sara", "Ahmed"]
marks = [85, 90, 78]

print(combine_lists(names, marks))
```

Q5. Iterators & StopIteration

```python

# Q5: Iterator and Exception Handling

numbers = [10, 20, 30, 40]

iterator = iter(numbers)

while True:
    try:
        print(next(iterator))
    except StopIteration:
        print("End of list reached.")
        break
```

Q6. Palindrome Using Slicing

```python

# Q6: Palindrome Checking

def is_palindrome(text):
    return text == text[::-1]

word = "madam"

print("Is Palindrome:", is_palindrome(word))
```

Q7. Second Largest Number Without Sorting

```python

# Q7: Second Largest Number Without Sorting

def second_largest(numbers):
    largest = None
    second = None

    for num in numbers:
        if largest is None or num > largest:
            second = largest
            largest = num
        elif num != largest and (second is None or num > second):
            second = num

    if second is None:
        raise ValueError("At least two distinct numbers are required")

    return second

numbers = [10, 25, 8, 17, 20]

print("Second Largest Number:", second_largest(numbers))
```
